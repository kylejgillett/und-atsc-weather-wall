```python
from PIL import Image
import glob
import os

# -----------------------
# User-tunable parameters
# -----------------------
max_size = (900, 900)   # max width/height in pixels
colors = 256            # try 128 if still too big
frame_skip = 1          # keep every Nth frame
frame_duration = 900    # ms per frame

# -----------------------
# Load and process frames
# -----------------------
frames = []

imgs = sorted(glob.glob('nexrad_*.png'))
print(f"Found {len(imgs)} images")

for fname in imgs[::frame_skip]:
    img = Image.open(fname).convert("RGBA")

    # Resize while preserving aspect ratio
    img.thumbnail(max_size, Image.LANCZOS)

    # Convert to palette-based image for GIF compression
    img = img.convert(
        "P",
        palette=Image.ADAPTIVE,
        colors=colors
    )

    frames.append(img)

print(f"Using {len(frames)} frames after skipping")

# -----------------------
# Save optimized GIF
# -----------------------
output = "08072025_interaction_loop.gif"

frames[0].save(
    output,
    format="GIF",
    append_images=frames[1:],
    save_all=True,
    duration=frame_duration,
    loop=0,            # infinite loop (Twitter-friendly)
    optimize=True,
    disposal=2
)

# -----------------------
# Report file size
# -----------------------
size_mb = os.path.getsize(output) / 1e6
print(f"Saved {output} ({size_mb:.2f} MB)")
```
